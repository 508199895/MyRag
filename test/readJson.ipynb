```python
import json

file1 = r"E:\007.agent\007.project\RAG\data\BEIR-NQ\nq\queries.jsonl"
file2 = r"E:\007.agent\007.project\RAG\data\BEIR-NQ\nq\corpus.jsonl"

with open(file1, "r", encoding="utf-8") as f1, open(file2, "r", encoding="utf-8") as f2:
    for i, (line1, line2) in enumerate(zip(f1, f2)):
        if i >= 5:
            break

        obj1 = json.loads(line1)
        obj2 = json.loads(line2)

        print(f"第 {i+1} 对")
        print("file1:")
        print(json.dumps(obj1, ensure_ascii=False, indent=2))
        print("file2:")
        print(json.dumps(obj2, ensure_ascii=False, indent=2))
        print("=" * 60)
```
